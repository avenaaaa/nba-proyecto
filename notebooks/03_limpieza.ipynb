# 03 · Limpieza

**Pregunta de este notebook:** ¿están los datos listos para responder las cuatro preguntas del proyecto?

**Entrada:** la copia raw (`data/raw/team_game_log_2025_26.parquet`), que no se modifica.
**Salida:** una tabla limpia en `data/clean/`.

**Qué hago:** aplico las ocho decisiones del notebook 02, en este orden:
1. Quedarme con las columnas útiles
2. Convertir la fecha
3. Crear las columnas nuevas
4. Validar
5. Guardar la tabla limpia

```python
import pandas as pd

raw = pd.read_parquet("../data/raw/team_game_log_2025_26.parquet")

columnas = [
    "GAME_ID", "GAME_DATE", "TEAM_ID", "TEAM_ABBREVIATION", "MATCHUP", "WL", "MIN",
    "FGM", "FGA", "FG_PCT", "FG3M", "FG3A", "FG3_PCT", "FTM", "FTA", "FT_PCT",
    "OREB", "DREB", "REB", "AST", "STL", "BLK", "TOV", "PF", "PTS", "PLUS_MINUS",
]

df = raw[columnas].copy()
df.shape
```

La tabla de trabajo conserva las 2.460 filas y 26 de las 29 columnas. Quedan fuera:

SEASON_ID: vale 22025 en todas las filas, porque el análisis cubre una sola temporada. Una columna con un único valor no distingue nada.
TEAM_NAME: repite la información de TEAM_ABBREVIATION.
VIDEO_AVAILABLE: no aporta a ninguna de las preguntas del análisis.

Las columnas se eligen con una lista de las que sí se usan, para que ninguna columna nueva de la API entre sin que lo decida.

## 2. Convertir GAME_DATE a fecha

```python
df["GAME_DATE"] = pd.to_datetime(df["GAME_DATE"], format="%Y-%m-%d")

print("Tipo:", df["GAME_DATE"].dtype)
print("Primer partido:", df["GAME_DATE"].min())
print("Último partido:", df["GAME_DATE"].max())
```
