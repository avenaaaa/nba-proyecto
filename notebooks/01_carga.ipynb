# 01 · Carga de datos
Descargo la temporada regular 2025-26 (una fila por equipo y partido) y guardo una copia raw que no se edita.

```python
from pathlib import Path
import pandas as pd
from nba_api.stats.endpoints import leaguegamelog

RAW = Path("../data/raw/team_game_log_2025_26.parquet")
RAW.parent.mkdir(parents=True, exist_ok=True)   # crea data/raw si no existe

if not RAW.exists():  # solo llama a la API si todavía no hay copia guardada
    respuesta = leaguegamelog.LeagueGameLog(
        season="2025-26",
        season_type_all_star="Regular Season",
        player_or_team_abbreviation="T",
        timeout=60,
    )
    respuesta.get_data_frames()[0].to_parquet(RAW, index=False)

df = pd.read_parquet(RAW)
df.shape
```
